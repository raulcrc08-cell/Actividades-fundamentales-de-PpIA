Ejercicio 1: Evaluador rendimiento de modelos

```python
def evaluar_modelo():
    try:
        # Se solicita el valor de la exactitud al usuario
        entrada = input("Ingresa la exactitud del modelo (un número entre 0.0 y 1.0): ")
        exactitud = float(entrada)

        # Y validamos que el número esté dentro del rango válido
        if exactitud < 0.0 or exactitud > 1.0:
            print("Error: El valor ingresado debe estar estrictamente entre 0.0 y 1.0.")
            return

        # Estructura condicional para categorizar el rendimiento
        if exactitud >= 0.90:
            categoria = "Excelente"
        elif exactitud >= 0.75:
            categoria = "Aceptable"
        elif exactitud >= 0.50:
            categoria = "Requiere Ajuste"
        else:
            categoria = "Crítico / No apto"

        # Se muestra el resultado
        print(f"\nExactitud registrada: {exactitud}")
        print(f"Categoría del modelo: {categoria}")

        return categoria

    except ValueError:
        print("Error: Por favor, introduce un valor numérico válido (ej. 0.85).")

evaluar_modelo()
```

Ejercicio 2: Procesamiento de Hiperparametros y Lotes

```python
# 1. Definición de la tupla global con valores fijos de arquitectura
CONFIGURACION_BASE = (512, 10)

# 2. Creación de la función que procesa las tasas de aprendizaje
def procesar_tasas_aprendizaje(tasas_lr):
    """
    Recibe una lista de tasas de aprendizaje, las escala mediante un operador
    aritmético y retorna la nueva lista junto con la configuración base.
    """
    factor_escala = 2.0
    tasas_escaladas = []

    # 3. Bucle for para iterar, operar y almacenar en la nueva lista
    for tasa in tasas_lr:
        tasa_modificada = tasa * factor_escala
        tasas_escaladas.append(tasa_modificada)

    # 4. aqui tenemos la impresión de los resultados solicitados
    print("--- REPORTE DE HIPERPARÁMETROS ---")
    print(f"Configuración Base (Tupla inmutable): {CONFIGURACION_BASE}")
    print(f"Tasas de aprendizaje originales: {tasas_lr}")
    print(f"Tasas de aprendizaje escaladas (x{factor_escala}): {tasas_escaladas}")

    # aqui retonarmos ambos valores
    return tasas_escaladas, CONFIGURACION_BASE

# --- Prueba de la interacción ---
tasas_ejemplo = [0.1, 0.01, 0.001, 0.0001]
procesar_tasas_aprendizaje(tasas_ejemplo)
```

Ejercicio 3: Simulador de Convergencia por epocas

```python
def simular_entrenamiento():
    # 1. Pedir el número máximo de épocas
    max_epocas = int(input("Ingresa el número máximo de épocas a simular: "))

    umbral = 0.05          # Tolerancia para el criterio de parada
    perdida = 1.0          # Pérdida inicial
    convergio = False

    # 2. Bucle for desde la época 1 hasta el límite indicado
    for epoca in range(1, max_epocas + 1):
        # 3. Pérdida decreciente usando la variable de control del bucle
        perdida = 1.0 / (epoca ** 1.5)

        print(f"Época {epoca}/{max_epocas} - Pérdida: {perdida:.4f}")

        # 4. Verificar si la pérdida cae por debajo del umbral
        if perdida < umbral:
            print(f"\nEl modelo convergió anticipadamente en la época {epoca} (pérdida < {umbral}).")
            convergio = True
            break

    if not convergio:
        print(f"\nEl entrenamiento terminó las {max_epocas} épocas sin alcanzar el umbral de {umbral}.")

    return perdida


# Invocar la función para probar
simular_entrenamiento()
```
